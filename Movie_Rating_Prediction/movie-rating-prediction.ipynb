```python
#import required libraries
import pandas as pd #to read dataset
import numpy as np #numeric work
import matplotlib.pyplot as plt #graph 
```

```python
#create a simple movie dataset
data=pd.DataFrame({
    'Budget':[50,60,70,80,90,40,30,100,110,120],
    'Popularity':[60,65,70,75,80,55,50,85,90,95],
    'Duration':[120,130,125,140,150,110,100,160,170,180],
    'Rating':[6.5,7.0,7.2,7.8,8.0,6.0,5.8,8.5,8.8,9.0]
})
#budget,popularity,duration are inputs
#rating is output
```

```python
#select input
features=data[['Budget','Popularity','Duration']]
#select target
target=data['Rating']
```

```python
from sklearn.model_selection import train_test_split

X_train_data,X_test_data,y_train_data,y_test_data = train_test_split(
    features,target,test_size=0.2,random_state=42)
#training and testing
```

```python
from sklearn.linear_model import LinearRegression

#create model
model=LinearRegression()

#train model
model.fit(X_train_data,y_train_data)
```

```python
#predict movie rating
predicted_ratings=model.predict(X_test_data)
```

```python
from sklearn.metrics import mean_squared_error

#calculate error
mse=mean_squared_error(y_test_data,predicted_ratings)
print("mean squared error:",mse)
```

```python
#plot actual vs predicted rating
plt.figure()
plt.scatter(y_test_data,predicted_ratings)
plt.xlabel("actual rating")
plt.ylabel("predicted ratings")
plt.title("actualvs predicted movie ratings")
plt.show()
```
